Prueba de bondad de ajuste

```python
import pandas as pd
import matplotlib.pyplot as plt
import statsmodels.api as sm
import scipy.stats as stats
```

```python
df = pd.read_excel('/content/drive/MyDrive/Colab Notebooks/projects5to/data/Tiempo_clientes(2).xlsx')
df
```

```python
df.hist(column = ["Tiempo entre llegadas (min)"])
```

```python
promedio = df["Tiempo entre llegadas (min)"].mean()
```

```python
#gráfica QQ-plot
sm.qqplot(df["Tiempo entre llegadas (min)"], stats.expon, line = "45", scale = promedio)
plt.show()
```

```python
#test kolmogorov
stats.kstest(df["Tiempo entre llegadas (min)"], "expon", args = (0, promedio))
```
